原始账本保持不变。该面板只批准数值层，模型资格仍为False。空白分项不填零，2024决算与汇编版本并列。


```python
from pathlib import Path
import csv, json, hashlib
from collections import Counter
root = Path.cwd()
if not (root / 'R_numeric_verification.csv').exists():
    root = root / 'research' / 'data' / '财政收入核准_20261010'
def read(name):
    with (root / name).open(encoding='utf-8-sig', newline='') as f:
        return list(csv.DictReader(f))
rows = read('R_numeric_verification.csv')
panel = read('R_province_year_numeric_panel.csv')
assert len(rows) == 906
assert len(panel) == 930
assert len({(r['year'], r['province']) for r in panel}) == 930
assert sum(bool(r['R_adjudicated_yi']) for r in panel) == 662
assert all(r['model_eligible'] == 'False' for r in panel)
print(Counter(r['numeric_status'] for r in panel))
```
